```python
from langgraph.graph import StateGraph, START, END
from langchain_openai import ChatOpenAI
from typing import TypedDict
from dotenv import load_dotenv
load_dotenv()  # take environment variables from .env file  

from langgraph.checkpoint.memory import InMemorySaver
```

```python
class State(TypedDict):
    topic: str
    joke : str
    explanation : str
```

```python
model = ChatOpenAI(temperature=0)
```

```python
def generate_joke(state: State):
    prompt = f"Tell me a joke on the topic: {state['topic']}"
    response = model.invoke(prompt).content
    return {'joke': response}
```

```python
def generate_explanation(state: State):
    prompt = f"Explain the following joke: {state['joke']}"
    response = model.invoke(prompt).content
    return {'explanation': response}
```

```python
graph = StateGraph(State)

checkpoint_saver = InMemorySaver()

graph.add_node("gen_joke", generate_joke)
graph.add_node("gen_explanation", generate_explanation)

graph.add_edge(START, "gen_joke")
graph.add_edge("gen_joke", "gen_explanation")
graph.add_edge("gen_explanation", END)

workflow = graph.compile(checkpointer=checkpoint_saver)

workflow
```

```python
thread_id = 1
config1 = {'configurable':{'thread_id': thread_id}}

initial_state1 = {'topic': 'work from home'}

final_state1 = workflow.invoke(initial_state1, config=config1)

final_state1
```

```python
workflow.get_state(config1)
```
